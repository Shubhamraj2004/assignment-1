Question 1 -Question 1: What is the role of filters and feature maps in Convolutional Neural 
Network (CNN)? 

Answer - Role of Filters in CNN
Filters are small matrices of learnable weights that slide over the input data (such as an image) during the convolution operation. Each filter is designed during training to detect specific patterns in the data, such as edges, corners, or textures. By applying many filters, a CNN can detect a diverse set of features from the input. Filters are responsible for capturing local patterns, and their weights are learned automatically to maximize task performance. Importantly, filters enable CNNs to recognize features regardless of the feature’s location in the input, which leads to translation invariance.

Purpose of Feature Maps
A feature map (also called an activation map) is the result of applying a single filter across the full input. Each feature map represents the spatial response of its corresponding filter—showing where the pattern the filter detects appears in the input. In the early layers, feature maps may capture basic visual features like edges or simple shapes, while deeper layers combine information from many filters to encode higher-level structures, such as object parts or full objects. The number of feature maps in a network determines its ability to learn and represent a variety of patterns and is a key hyperparameter.

Question 2: Explain the concepts of padding and stride in CNNs(Convolutional Neural 
Network). How do they affect the output dimensions of feature maps?

Answer -Padding in CNNs
Padding refers to adding extra pixels (usually zeros) around the border of the input image before applying the convolution filter. The main reasons for padding are:

Preserving spatial dimensions: "Same" padding adds just enough pixels so the output feature map has the same size as the input; "Valid" padding uses no padding, so the output shrinks with each convolution.

Using border information: Padding ensures that edge pixels are included in the convolution instead of losing information at the borders.

Stride in CNNs
Stride defines how many pixels the filter jumps with each move. For example:

Stride 1: Filter slides one pixel at a time.

Stride 2: Filter jumps by two pixels each move, covering less of the image and reducing the output map size faster.

Question 3: Define receptive field in the context of CNNs. Why is it important for deep 
architectures?

Answer - The receptive field in a Convolutional Neural Network (CNN) is the region of the input image that a particular neuron or feature in a given network layer is able to "see" or be influenced by. As information moves deeper through the network’s layers, receptive fields grow larger—layer by layer, each neuron’s output depends on a broader region of the original image.

Key Points
In early layers, each neuron is affected by a small, local patch of the input, such as just a few neighboring pixels.

In deeper layers, receptive fields expand because each neuron pools information from an increasingly wide area of the input, effectively summarizing more complex and abstract information.

The receptive field size determines how much context the CNN can utilize for making decisions about the content in the image.

Importance for Deep Architectures
Larger receptive fields are critical in deep networks; they enable the model to capture complex structures and global context, rather than being limited to local details.

For tasks like object recognition, segmentation, or scene understanding, neurons in later layers must access enough context to identify larger features or entire objects.

If the receptive field is too small, deep CNNs might miss relationships between important regions or detect only fragmentary patterns, leading to suboptimal understanding or predictions.

Techniques like pooling and using larger kernel sizes, or dilated convolutions, are used to increase the receptive field efficiently.

In summary, a well-designed receptive field ensures that each output neuron considers enough spatial context, which is especially important in deep CNNs for understanding complex patterns and making accurate predictions.

Answer- Calculating number of parameters in Convolutional Neural Networks (CNNs) is important for understanding model complexity, computational requirements and potential overfitting. Parameters in CNNs are primarily weights and biases learned during training.

Steps to Calculate Number of Parameter in CNN
To calculate the total number of parameters in a 2D convolutional neural network which includes convolutional, fully connected and batch normalization layers while excluding pooling layers as they contribute zero parameters. There steps are:

1. Convolutional Layer
Each convolutional layer has filters or kernels that scan the input image or feature map.

Here the "+ 1" accounts for the bias term for each filter.

Let's consider an example: if a layer has 32 filters of size 3x3 and input with 3 channels (RGB image). Then its parameters will be: (3×3×3+1)×32=(27+1)×32=28×32=896 parameters

2. Fully Connected (Dense) Layers
For a fully connected layer, the number of parameters is given by the number of input units times the number of output units, plus one bias term for each output unit.


Let's take an example: if the fully connected layer has 128 input units and 64 output units the computed parameters are: (128×64)+64=8192+64=8256 Parameters.

3. Batch Normalization Layers
For batch normalization layers each feature channel has two parameters (gamma and beta).


Let's take an example: Suppose that batch normalization layer is applied to an output of 64 channels then parameters can be computed as: 2 x 64 =128 parameters.

4. Pooling Layers
Pooling layers like max pooling, average pooling, etc do not have learnable parameters so they contribute 0 to the parameter count.

5. Combining All Layers
To find the total number of parameters in the CNN sum all the parameters from all the layers calculated above.

Example of Calculating the Number of Parameter in CNN
Consider a simple CNN with the following layers:

Conv layer: 16 filters, 3x3 size, 3 input channels
Conv layer: 32 filters, 3x3 size, 16 input channels
Fully connected layer: 128 input units, 64 output units
Batch normalization after each convolutional layer
Below are the parametrs calculation:

Conv Layer 1: (3×3×3+1)×16=(27+1)×16=448
Batch Norm 1: 2×16=32
Conv Layer 2: (3×3×16+1)×32=(144+1)×32=4640
Batch Norm 2: 2×32=64
Fully Connected Layer: (128×64)+64=8256
Total Parameters: 448+32+4640+64+8256=13440
So, the total number of parameters in this simple CNN example is 13,440.

Parameter Calculation for 3-D Convolutions
For a 3D convolutional layer, the number of parameters depends on the size of the filters (kernels), number of filters and the number of input channels.

Question 5. Compare and contrast different CNN-based architectures like LeNet, 
AlexNet, and VGG in terms of depth, filter sizes, and performance. 

Answer -. LeNet (1998)
Depth: ~5–7 layers (very shallow by today’s standards).

Filter Sizes:

Conv layers: 5×5 filters

Pooling: average pooling (subsampling)

Performance:

Designed for digit recognition (MNIST).

First successful CNN architecture.

Low computational cost (good for the hardware of the 1990s).

Limited generalization to complex datasets.

🔹 2. AlexNet (2012)

Depth: 8 layers (5 conv + 3 fully connected).

Filter Sizes:

1st Conv: 11×11 filters, stride 4 (large receptive field).

Later convs: 5×5, then 3×3 filters.

Pooling: max pooling instead of average pooling.

Performance:

Won ImageNet 2012 with a huge margin (top-5 error dropped from 26% → 15%).

Introduced ReLU activations (faster training).

Used dropout + data augmentation to reduce overfitting.

Leveraged GPU acceleration, making deep CNNs practical.

🔹 3. VGG (2014)

Depth: 16–19 layers (VGG-16, VGG-19).

Filter Sizes:

Only 3×3 filters stacked in depth (instead of large filters like 11×11).

Pooling: 2×2 max pooling.

Used the principle: multiple small filters stacked = larger receptive field with fewer parameters.

Performance:

Very strong ImageNet results (2nd place, 2014).

More uniform, elegant design compared to AlexNet.

High computational + memory cost (lots of parameters).

Became the foundation for many later models (ResNet, MobileNet, etc.).

```python
# Question 6. Using keras, build and train a simple CNN model on the MNIST dataset from scratch. Include code for module creation, compilation, training, and evaluation. 
# Step 1: Import libraries
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# Step 2: Load and preprocess MNIST dataset
(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()

# Reshape to (28,28,1) for CNN input
x_train = x_train.reshape(-1, 28, 28, 1).astype("float32") / 255.0
x_test = x_test.reshape(-1, 28, 28, 1).astype("float32") / 255.0

# Step 3: Build CNN model
model = models.Sequential([
    layers.Conv2D(32, (3,3), activation="relu", input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation="relu"),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")   # 10 digits
])

# Step 4: Compile model
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

# Step 5: Train model
history = model.fit(x_train, y_train, epochs=5,
                    batch_size=64,
                    validation_split=0.1)

# Step 6: Evaluate model
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"Test accuracy: {test_acc:.4f}")
```

```python
#Question 7: Load and preprocess the CIFAR-10 dataset using Keras, and create a CNN model to classify RGB images. Show your preprocessing and architecture
#Answer


# Step 1: Import Libraries
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt
import numpy as np

# Step 2: Load CIFAR-10 Dataset
(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()
print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)

# Step 3: Normalize Pixel Values (0–1)
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# Step 4: One-hot Encode Labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Step 5: Define CNN Model Architecture
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation='relu'),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

# Step 6: Compile Model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Step 7: Train Model
history = model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=64,
    validation_data=(x_test, y_test),
    verbose=1
)

# Step 8: Evaluate Model
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"\n✅ Test Accuracy: {test_acc * 100:.2f}%")

# Step 9: Plot Accuracy and Loss
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

# Step 10: Display Some Test Predictions
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

predictions = model.predict(x_test)

plt.figure(figsize=(10,4))
for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(x_test[i])
    plt.title(f"Pred: {class_names[np.argmax(predictions[i])]}")
    plt.axis('off')
plt.show()
```

```python
#Question 8: Using PyTorch, write a script to define and train a CNN on the MNIST dataset. Include model definition, data loaders, training loop, and accuracy evaluation.
#Answer
# ✅ CNN on MNIST Dataset using PyTorch

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import torch.nn.functional as F

# Step 1: Device Configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Step 2: Data Loading and Preprocessing
transform = transforms.Compose([
    transforms.ToTensor(),  # Convert images to tensor
    transforms.Normalize((0.1307,), (0.3081,))  # Normalize with MNIST mean & std
])

train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root='./data', train=False, download=True, transform=transform)

train_loader = DataLoader(dataset=train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(dataset=test_dataset, batch_size=1000, shuffle=False)

# Step 3: Define CNN Model
class CNNModel(nn.Module):
    def __init__(self):
        super(CNNModel, self).__init__()
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(64 * 7 * 7, 128)
        self.fc2 = nn.Linear(128, 10)
        self.dropout = nn.Dropout(0.5)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(-1, 64 * 7 * 7)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x

model = CNNModel().to(device)
print(model)

# Step 4: Define Loss Function and Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Step 5: Training Loop
num_epochs = 5
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        
        optimizer.zero_grad()        # Reset gradients
        outputs = model(images)      # Forward pass
        loss = criterion(outputs, labels)  # Compute loss
        loss.backward()              # Backpropagation
        optimizer.step()             # Update weights
        
        running_loss += loss.item()
    
    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {running_loss/len(train_loader):.4f}")

# Step 6: Accuracy Evaluation
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total
print(f"\n✅ Test Accuracy: {accuracy:.2f}%")
```
