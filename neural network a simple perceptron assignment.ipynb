Question 1. what is deep learning?Briefly describe how it evolved and how it differs from traditional machine learning.
Answer -Deep learning is a specialized subset of machine learning that leverages multilayered artificial neural networks to autonomously extract patterns from complex, high-dimensional data and make intelligent decisions, mimicking the processing power of the human brain.

Evolution of Deep Learning
Deep learning emerged as an advancement of traditional machine learning, with significant growth in the last decade driven by increased computational power, large datasets, and innovations in neural network architectures. Early machine learning models relied on simpler algorithms and feature engineering, but advancements in neural network designs enabled models to learn hierarchical representations, making deep learning especially effective for tasks such as image and speech recognition, natural language processing, and more.

Key Differences from Traditional Machine Learning
Feature Engineering
Traditional machine learning depends heavily on manual feature engineering and domain expertise to extract useful attributes from data before feeding them to algorithms like decision trees, linear regression, or SVMs. Deep learning automates this process, enabling neural networks to learn and extract relevant features directly from raw, unstructured data through multiple layers.

Data Requirements
Machine learning models often work well with smaller datasets and structured information, whereas deep learning requires vast amounts of data for effective training and improved performance, especially with complex inputs.

Model Complexity and Tasks
Deep learning networks—such as CNNs, RNNs, and MLPs—handle highly complex, non-linear relationships and excel at unstructured data like images, audio, and text, whereas traditional models perform better with simpler, structured problems.

Computational Power
Deep learning demands substantial computational resources, typically requiring GPUs or TPUs to process large volumes of data and train deep neural networks. Traditional machine learning models are less computationally intensive and can often be trained using standard hardware.

Interpretability
Traditional machine learning models are usually more interpretable, making it easier to understand their decision processes. Deep learning models—often described as "black boxes"—are more challenging to interpret despite ongoing efforts to increase transparency.
Question 2: Explain the basic architecture and functioning of a Perceptron. What are its 
limitations? 
Answer -A Perceptron is the simplest type of artificial neural network, designed for binary classification, and is often considered the foundational building block for more complex neural networks and deep learning architectures.

Basic Architecture
Inputs: The perceptron receives one or more input signals, each representing a feature of the data.

Weights: Every input is associated with a weight value that signifies its importance.

Summation: The inputs are multiplied by their respective weights, summed, and a bias term may be added.

Activation Function: The sum is passed through an activation function (typically a step or sign function) that determines the output—usually 0 or 1 (binary).

Output: The activation function produces the final output, classifying the input as belonging to one of two classes.

Functioning
During training, the perceptron adjusts its weights and bias using the perceptron learning rule to minimize errors between the predicted and true outputs.

If the weighted sum exceeds a threshold, the output is 1; otherwise, it's 0.

Limitations
Only Linearly Separable Data: The perceptron can only solve problems where the classes are linearly separable—that is, data that can be separated by a straight line or hyperplane. It cannot learn XOR-like or other non-linear problems.

Binary Output: The classic perceptron delivers only binary output (0 or 1). It can't handle multi-class or regression problems natively.

No Deep Structure: It consists of just one layer and lacks the capacity to learn hierarchical or abstract patterns, hindering its effectiveness for complex problems.

Limited Flexibility: More complex or nuanced classifications require multi-layer networks (multi-layer perceptrons), for which backpropagation and different activation functions are typically used.

The perceptron's simplicity opened the door for neural network research, but its limitations motivated the development of multi-layer architectures and more advanced learning algorithms.

Question 4. What is the difference between Loss function and Cost function in neural 
networks? Provide examples. 
Answer -The difference between a loss function and a cost function in neural networks lies in their scope and how they are used to quantify errors during training.

Loss Function
The loss function measures the error or deviation between a model’s prediction and the actual value for a single training example.

It is calculated for each data point and expresses how well the model performs on that one instance (e.g., image, sample, record).


Cost Function
The cost function is an aggregate measure, summarizing the overall error of the model on the entire dataset (or batch of samples) by aggregating the individual losses—commonly through their mean or sum.

It provides a scalar value used for optimization during training (minimizing cost gives better model performance).
Question 5: What is the role of optimizers in neural networks? Compare Gradient 
Descent, Adam, and RMSprop.  
Answer .Optimizers in neural networks play a critical role in training by iteratively adjusting the network’s weights and biases to minimize the loss function, thereby improving the accuracy and performance of the model. They guide how the parameters are updated during backpropagation, facilitating efficient learning and convergence to an optimal or near-optimal solution.

Role of Optimizers
Optimize model parameters (weights and biases) to minimize the difference between predicted output and true output (loss function).

Control how the learning happens by determining the step size and direction of parameter updates (learning rate and gradient evaluation).

Help avoid local minima and improve convergence speed and stability through various strategies (momentum, adaptive learning rate).
Optimizers in neural networks are algorithms that update the model parameters (weights and biases) to minimize the loss function and improve model accuracy. Their role is to guide the learning process by deciding how and how much to adjust parameters based on the gradients calculated during backpropagation.

Gradient Descent is the most basic optimizer. It updates weights by moving in the direction opposite to the gradient of the loss function. While simple and foundational, it can be slow to converge and is sensitive to the choice of learning rate. It may also get stuck in local minima or saddle points, which limits its efficiency in complex models.

Adam (Adaptive Moment Estimation) is a more advanced optimizer that combines ideas from both momentum optimization and adaptive learning rate methods. It keeps track of the first and second moments (mean and variance) of the gradients to adaptively tune learning rates for each parameter. This results in faster convergence and better handling of noisy or sparse gradients, making Adam highly popular for large and complex datasets. However, it can sometimes overfit or converge to suboptimal solutions if not properly tuned.

RMSprop (Root Mean Square Propagation) is another adaptive learning rate method that divides the learning rate by a moving average of recent gradient magnitudes, normalizing parameter updates. This helps it perform well on problems with non-stationary objectives, such as training recurrent neural networks. RMSprop is generally more stable than plain gradient descent but requires careful hyperparameter tuning, and it may sometimes overshoot minima.




```python
#Question 6: Write a Python program to implement a single-layer perceptron from scratch using NumPy to solve the logical AND gate.
#Answer
import numpy as np

class Perceptron:
    def __init__(self, input_size, learning_rate=0.1, epochs=10):
        self.weights = np.zeros(input_size + 1)  # +1 for bias
        self.learning_rate = learning_rate
        self.epochs = epochs

    def activation(self, x):
        # Step activation function
        return 1 if x >= 0 else 0

    def predict(self, x):
        # Calculate weighted sum plus bias
        z = np.dot(x, self.weights[1:]) + self.weights[0]
        return self.activation(z)

    def train(self, X, y):
        for epoch in range(self.epochs):
            for inputs, label in zip(X, y):
                prediction = self.predict(inputs)
                # Update weights and bias
                self.weights[1:] += self.learning_rate * (label - prediction) * inputs
                self.weights[0] += self.learning_rate * (label - prediction)  # bias update

# Define dataset for AND gate
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([0, 0, 0, 1])

# Create perceptron object
perceptron = Perceptron(input_size=2, learning_rate=0.1, epochs=10)

# Train perceptron
perceptron.train(X, y)

# Test perceptron
print("Predictions for AND gate:")
for x_input in X:
    print(f"Input: {x_input} -> Prediction: {perceptron.predict(x_input)}")
```

```python
#Question 7: Implement and visualize Sigmoid, ReLU, and Tanh activation functions using Matplotlib.
#Answer
import numpy as np
import matplotlib.pyplot as plt

# Define input range
x = np.linspace(-10, 10, 400)

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# ReLU function
def relu(x):
    return np.maximum(0, x)

# Tanh function
def tanh(x):
    return np.tanh(x)

# Compute activation outputs
sigmoid_y = sigmoid(x)
relu_y = relu(x)
tanh_y = tanh(x)

# Plotting the activation functions
plt.figure(figsize=(10, 6))
plt.plot(x, sigmoid_y, label='Sigmoid', color='blue')
plt.plot(x, relu_y, label='ReLU', color='green')
plt.plot(x, tanh_y, label='Tanh', color='red')

plt.title('Activation Functions: Sigmoid, ReLU, Tanh')
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Question 8: Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy. 
#Answer

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize inputs to range [0, 1]
x_train = x_train / 255.0
x_test = x_test / 255.0

# Build the model
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

# Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
history = model.fit(x_train, y_train, epochs=5, batch_size=32, verbose=1)

# Print final training accuracy
print(f"Final training accuracy: {history.history['accuracy'][-1]:.4f}")
```

```python
#Question 9. Visualize the loss and accuracy curves for a neural network model trained on the Fashion MNIST dataset. Interpret the training behavior. 
#Answer
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
import matplotlib.pyplot as plt

# Load the Fashion MNIST dataset
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Normalize pixel values
train_images = train_images / 255.0
test_images = test_images / 255.0

# Build a simple neural network model
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train the model with validation split
history = model.fit(train_images, train_labels, epochs=25, validation_split=0.2)

# Plot the loss curve
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper right')
plt.show()

# Plot the accuracy curve
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='lower right')
plt.show()
```

Question 10: You are working on a project for a bank that wants to automatically detect 
fraudulent transactions. The dataset is large, imbalanced, and contains structured 
features like transaction amount, merchant ID, and customer location. The goal is to 
classify each transaction as fraudulent or legitimate. 
Explain your real-time data science workflow: 
● How would you design a deep learning model (perceptron or multilayer NN)? 
● Which activation function and loss function would you use, and why? 
● How would you train and evaluate the model, considering class imbalance? 
● Which optimizer would be suitable, and how would you prevent overfitting? 
Answer - A real-time workflow for deep learning-based fraud detection on large, imbalanced, structured banking data involves several important steps and best practices.

Model Design: Architecture Choices
Start with a multilayer perceptron (MLP) or deep neural network, as structured data can be handled effectively by fully connected layers.

Input: Transaction features such as amount, merchant ID, customer location, encoded as numeric/categorical vectors.

Recommended design for an MLP:

Input layer matching feature dimension.

Multiple hidden layers (Dense) with 64–256 neurons, 'relu' activation for non-linearity.

Output layer: Single neuron (binary classification) or two neurons (softmax) for 'fraud' vs 'legitimate'.

Activation and Loss Functions: Choosing Wisely
Use ReLU activation in hidden layers for efficiency and gradient stability.

Use sigmoid activation for single-neuron output (binary) or softmax for two outputs (multiclass).

For imbalanced binary classification, use binary cross-entropy loss; it works well with sigmoid output.

To address imbalance, apply class weights so that the loss penalizes errors on the minority 'fraud' class more heavily.

Training and Evaluation Strategies
Train using class_weight in Keras/TensorFlow to compensate for class imbalance.

Evaluate using metrics beyond accuracy:

Precision, recall, F1 score, ROC AUC—these reflect the model's ability to catch fraudulent transactions without excessive false alarms.

Use stratified train-test split so both sets contain similar fraud class proportions.

Optimizer Choice and Overfitting Prevention
Optimizer: Adam is the most suitable—fast convergence, robust for sparse gradients, works well with large datasets.

Prevent overfitting with these strategies:

Dropout layers in the network.

Early stopping during training based on validation loss.

Regularization (l2, l1 penalties on weights).

Monitor validation metrics as training progresses to ensure generalization.
