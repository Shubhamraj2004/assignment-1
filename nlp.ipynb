Question 2: What is TextBlob and how does it simplify common NLP tasks like
sentiment analysis and translation?
Answer - TextBlob is a simple and user-friendly Python library built on top of NLTK and Pattern.
Its goal is to make common NLP tasks easier by providing a clean, high-level API so you don’t have to deal with complex preprocessing or algorithms.

It is designed mainly for:

Beginners in NLP

Rapid prototyping

Applications that need quick text processing without heavy ML models
1. Sentiment Analysis

TextBlob provides an in-built sentiment analyzer.
With just one line of code, you get:

Polarity → range: −1 (negative) to +1 (positive)

Subjectivity → range: 0 (objective) to 1 (subjective)

Why it's simpler?

No need to train your own model

No need to tokenize or clean text manually

Direct function gives results instantly
2. Translation

TextBlob integrates with online translation APIs (like Google Translate).
You can translate text in one line without manually calling APIs or handling authentication.

Question 4: Describe the architecture and functioning of a Recurrent Natural Network
(RNN).

Answer- Recurrent Neural Network (RNN) is a type of neural network designed to handle sequential data such as text, speech, time-series, and sensor data.
Unlike feed-forward networks, RNNs have a memory that stores information from previous time steps.

 1. Architecture of an RNN

An RNN consists of three main layers:

1️ Input Layer

Takes a sequence as input
(e.g., words in a sentence, values in a time-series)

Input at time t is denoted as xₜ

2️ Hidden Layer (with recurrent connections)

Core component of RNN

Has a loop that allows the network to pass information from one step to the next

Hidden state at time t: hₜ

Question 5: What is the key difference between LSTM and GRU networks in NLP
applications?

Answer - LSTM (Long Short-Term Memory) and GRU (Gated Recurrent Unit) are advanced RNN architectures designed to solve the vanishing gradient problem.
Both use gates to control information flow, but they differ in structure and complexity.

 1. Architectural Differences
LSTM Architecture

Has three gates:
✔ Input Gate
✔ Forget Gate
✔ Output Gate

Has two states:
✔ Cell State (long-term memory)
✔ Hidden State (short-term memory)

GRU Architecture

Has two gates:
✔ Update Gate
✔ Reset Gate

Has one state:
✔ Hidden State only

No separate cell state

 2. Functional Differences
LSTM

Separates long-term memory (cell state) and short-term memory (hidden state)

More powerful for capturing long-range dependencies

More parameters → heavier and slower

GRU

Combines cell state + hidden state into one

Fewer parameters → faster training and inference

Works well with smaller datasets

Often performs similarly to LSTM but with less computational cost

 3. Key Difference (One-Line Answer)

LSTM uses three gates and maintains both a hidden state and a separate cell state, making it more complex but better at long-range dependency learning, while GRU uses only two gates and a single hidden state, making it simpler, faster, and more efficient.


```python
# Question 6: Write a Python program using TextBlob to perform sentiment analysis on
# the following paragraph of text:
# “I had a great experience using the new mobile banking app. The interface is intuitive,
# and customer support was quick to resolve my issue. However, the app did crash once
# # during a transaction, which was frustrating"
from textblob import TextBlob

# Given paragraph
text = """
I had a great experience using the new mobile banking app.
The interface is intuitive, and customer support was quick to resolve my issue.
However, the app did crash once during a transaction, which was frustrating.
"""

# Create a TextBlob object
blob = TextBlob(text)

# Perform sentiment analysis
sentiment = blob.sentiment

# Display the results
print("Sentiment Analysis Result:")
print(f"Polarity: {sentiment.polarity}")       # Range: -1 (negative) to +1 (positive)
print(f"Subjectivity: {sentiment.subjectivity}")  # Range: 0 (objective) to 1 (subjective)
```

```python
# Question 7: Given the sample paragraph below, perform string tokenization and
# frequency distribution using Python and NLTK:
# “Natural Language Processing (NLP) is a fascinating field that combines linguistics,
# computer science, and artificial intelligence. It enables machines to understand,
# interpret, and generate human language. Applications of NLP include chatbots,
# sentiment analysis, and machine translation. As technology advances, the role of NLP
# in modern solutions is becoming increasingly critical.”
import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

# Download required resources (only once)
nltk.download('punkt')
nltk.download('punkt_tab') # Added this line based on the error message

# Given paragraph
text = """
Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical.
"""

# Step 1: Tokenization
tokens = word_tokenize(text)

# Step 2: Frequency Distribution
freq_dist = FreqDist(tokens)

# Display results
print("Tokens:")
print(tokens)
print("\nFrequency Distribution:")
for word, frequency in freq_dist.items():
    print(f"{word}: {frequency}")
```

```python
# Question 8: Implement a basic LSTM model in Keras for a text classification task using
# the following dummy dataset. Your model should classify sentences as either positive
# (1) or negative (0).
# # Dataset
# texts = [
# “I love this project”, #Positive
# “This is an amazing experience”, #Positive
# “I hate waiting in line”, #Negative
# “This is the worst service”, #Negative
# “Absolutely fantastic!” #Positive
# ]
# labels = [1, 1, 0, 0, 1]
# Preprocess the text, tokenize it, pad sequences, and build an LSTM model to train on
# this data. You may use Keras with TensorFlow backend.
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Dummy dataset
texts = [
    "I love this project",                   # Positive
    "This is an amazing experience",         # Positive
    "I hate waiting in line",                # Negative
    "This is the worst service",             # Negative
    "Absolutely fantastic!"                  # Positive
]

labels = [1, 1, 0, 0, 1]

# -----------------------------
# Step 1: Text Tokenization
# -----------------------------
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

# Convert text to sequences
sequences = tokenizer.texts_to_sequences(texts)

print("Sequences:", sequences)

# Vocabulary size
vocab_size = len(tokenizer.word_index) + 1
print("Vocabulary Size:", vocab_size)

# -----------------------------
# Step 2: Padding Sequences
# -----------------------------
max_len = max(len(seq) for seq in sequences)
padded_sequences = pad_sequences(sequences, maxlen=max_len, padding='post')

print("Padded Sequences:\n", padded_sequences)

# Convert labels to array
labels = np.array(labels)

# -----------------------------
# Step 3: Build LSTM Model
# -----------------------------
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=16, input_length=max_len))
model.add(LSTM(32, return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

# Compile model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Model summary
print(model.summary())

# -----------------------------
# Step 4: Train Model
# -----------------------------
model.fit(padded_sequences, labels, epochs=10, verbose=1)

# -----------------------------
# Step 5: Test Prediction
# -----------------------------
test_sentence = ["I really love this"]
test_seq = tokenizer.texts_to_sequences(test_sentence)
test_pad = pad_sequences(test_seq, maxlen=max_len)

prediction = model.predict(test_pad)
print("\nPrediction (Positive Probability):", prediction[0][0])
```

```python
# Question 9: Using spaCy, build a simple NLP pipeline that includes tokenization,
# lemmatization, and entity recognition. Use the following paragraph as your dataset:
# “Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the
# development of India’s atomic energy program. He was the founding director of the Tata
# Institute of Fundamental Research (TIFR) and was instrumental in establishing the
# Atomic Energy Commission of India.”
# Write a Python program that processes this text using spaCy, then prints tokens, their
# lemmas, and any named entities found.
import spacy

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# Given text
text = """
Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the
development of India’s atomic energy program. He was the founding director of the Tata
Institute of Fundamental Research (TIFR) and was instrumental in establishing the
Atomic Energy Commission of India.
"""

# Process the text
doc = nlp(text)

# ------ TOKENIZATION & LEMMATIZATION ------
print("Tokens and Lemmas:")
for token in doc:
    print(f"{token.text}  -->  {token.lemma_}")

# ------ NAMED ENTITY RECOGNITION ------
print("\nNamed Entities:")
for ent in doc.ents:
    print(f"{ent.text}  -->  {ent.label_}")
```
